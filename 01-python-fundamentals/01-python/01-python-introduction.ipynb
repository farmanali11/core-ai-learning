```python
def decorator_function(any_function):
  def wrapper_function():
    print("This is a Awesome Function")
    any_function()
  return wrapper_function

@decorator_function
def func1():
  print('This is a Function -01')
  func1()

@decorator_function
def func2():
  print("This is a function - 02")

func2()
```

```python
from functools import wraps
def decorator_function(any_function):
  def wrapper_function():
    @wraps(any_function)
    print("This is awesome function")
    return any_function(*args,**kwargs)
  return wrapper_function

@decorator_function
def add(a,b):
  return a+b
```
