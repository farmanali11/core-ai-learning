```python
import numpy as np

import pandas as pd

import math

import matplotlib

import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

def PointsInCircum(r,n=100):
  return [(math.cos(2*math.pi/n*x)*r+np.random.normal(-30,30),math.sin(2*math.pi/n*x)*r+np.random.normal(-30,30)) for x in range(1,n+1)]


```

```python
# 1. Generate all your data sources first
data1 = PointsInCircum(500, 1000)
data2 = PointsInCircum(300, 700)
data3 = PointsInCircum(100, 300)
data4 = [
    (np.random.randint(-600, 600), np.random.randint(-600, 600)) for i in range(300)
]

# 2. Convert them into DataFrames
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df3 = pd.DataFrame(data3)
df4 = pd.DataFrame(data4)

# 3. Combine them all at once (efficient!)
df = pd.concat([df1, df2, df3, df4], ignore_index=True)
```

```python
plt.figure(figsize=(6,6))

plt.scatter(df[0],df[1],s=15,color='grey')

plt.title('Datasets',fontsize=18)
plt.xlabel("Feature-01", fontsize=12)

plt.ylabel("Feature-02", fontsize=12)

plt.show()
```

```python
from sklearn.cluster import KMeans

```

```python
kmeans = KMeans(n_clusters=4,random_state=42)

kmeans.fit(df[[0,1]])

```

```python
df['KMeans_labels'] = kmeans.labels_
```

```python
colors = ['purple','red','blue','green']
plt.figure(figsize=(6,6))


plt.scatter(df[0], df[1], c=df['KMeans_labels'], cmap=matplotlib.colors.ListedColormap(colors), s=15)

plt.title("K Means Clustering Algorithm", fontsize=18)
plt.xlabel("Feature-01", fontsize=12)

plt.ylabel("Feature-02", fontsize=12)

plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

```

```python
model = AgglomerativeClustering(n_clusters=4, metric="euclidean", linkage="ward")

model.fit(df[[0, 1]])
```

```python
df['HR_labels'] = model.labels_
```

```python
colors = ["purple", "red", "blue", "green"]
plt.figure(figsize=(6, 6))


plt.scatter(
    df[0],
    df[1],
    c=df["HR_labels"],
    cmap=matplotlib.colors.ListedColormap(colors),
    s=15,
)

plt.title("Hiarerchacil Clustering Algorithm", fontsize=18)
plt.xlabel("Feature-01", fontsize=12)

plt.ylabel("Feature-02", fontsize=12)

plt.show()
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan = DBSCAN(eps=30,min_samples=6)

dbscan.fit(df[[0,1]])
```

```python
df['DBSCAN_labels'] = dbscan.labels_


df['DBSCAN_labels'] = dbscan.labels_

df['DBSCAN_labels'].value_counts()
```

```python
colors = ["purple", "red", "blue", "green"]
plt.figure(figsize=(6, 6))


plt.scatter(
    df[0],
    df[1],
    c=df["DBSCAN_labels"],
    cmap=matplotlib.colors.ListedColormap(colors),
    s=15,
)

plt.title("DBSCAN Clustering Algorithm", fontsize=18)
plt.xlabel("Feature-01", fontsize=12)

plt.ylabel("Feature-02", fontsize=12)

plt.show()
```
