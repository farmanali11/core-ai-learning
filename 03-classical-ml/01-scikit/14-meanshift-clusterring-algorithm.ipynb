```python
import numpy as np

import pandas as pd

import matplotlib.pyplot as plt

from sklearn.cluster import MeanShift

from sklearn.datasets import make_blobs

from mpl_toolkits.mplot3d import Axes3D
```

```python
cordinates = [[2,2,3],[6,7,8],[5,10,13]]


x,_ = make_blobs(n_samples=150,centers=cordinates,cluster_std=0.60)

```

```python
data_fig = plt.figure(figsize=(10,8))
ax = data_fig.add_subplot(111,projection='3d')

ax.scatter(x[:, 0], x[:, 1], x[:, 2],marker ='o',color='purple')

plt.show()
```

```python
from sklearn.cluster import estimate_bandwidth
```

```python
bandwith = estimate_bandwidth(x,quantile=0.2,n_samples=500)
```

```python
msc = MeanShift(bandwidth=bandwith,bin_seeding=True)
msc.fit(x)
clusters_center = msc.cluster_centers_

labels = msc.labels_

cluster_label= np.unique(labels)

n_clusters= len(cluster_label)

n_clusters
```

```python
msc_fig = plt.figure(figsize=(10, 8))
ax = msc_fig.add_subplot(111, projection="3d")

ax.scatter(
    clusters_center[:, 0],
    clusters_center[:, 1],
    clusters_center[:, 2],
    marker="o",
    color="green",
    s=300,
    linewidth=5,
    zorder=10,
)

ax.scatter(x[:, 0], x[:, 1], x[:, 2], marker="o", color="purple")


plt.title("Estimated Number of Clusters:%d" % n_clusters)


plt.show()
```
