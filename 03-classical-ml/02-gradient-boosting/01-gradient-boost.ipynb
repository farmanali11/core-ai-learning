```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn import datasets


data= datasets.load_diabetes()



```

```python
data 
```

```python
df =pd.DataFrame(data.data,columns=data.feature_names)
```

```python
df
```

```python
df['target'] = pd.Series(data.target)
```

```python
df
```

```python
df.info()
```

```python
df.duplicated()
```

```python
y=df['target']
X=df.drop('target',axis=1)
```

```python
from sklearn.model_selection import train_test_split


X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=0)
```

```python
from sklearn.ensemble import GradientBoostingClassifier


gbr = GradientBoostingClassifier(n_estimators=500)


gbr.fit(X_train,y_train)
```

```python
y_pred=gbr.predict(X_test)

print(y_pred)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_pred,y_test)

print(mae)
```

```python
feature_scores = pd.Series(gbr.feature_importances_,index=X_train.columns).sort_values(ascending=False)


print(feature_scores)
```

```python
sns.barplot(x=feature_scores,y=feature_scores.index)


plt.xlabel('Feature importance Score')

plt.ylabel('Feature')

```

```python
from cProfile import label

from numpy import dtype


test_score = np.zeros((500,),dtype=np.float64)


for i,y_pred in enumerate(gbr.staged_predict(X_test)):
  test_score[i] = np.mean((y_test - y_pred) ** 2)


fig = plt.figure(figsize=(10,10))

plt.subplot(1,1,1)


plt.title('Divergence')


plt.plot(np.arange(500)+1,gbr.train_score_,'b',label='Traing Set Divergence')
plt.plot(np.arange(500) + 1, gbr.train_score_, "r", label="Testing Set Divergence")


plt.legend(loc='upper right')

plt.xlabel('Boosting Iterations')

plt.ylabel('Deviance')

fig.tight_layout()

plt.show()
```
