```python
# 1. Target the unique total population icon
pop_img = soup.find('img', src=lambda x: x and 'icons/population.png' in x)

if pop_img:
    # 2. Go to the card container and grab the correct headline number
    parent_card = pop_img.find_parent('div', class_='card')
    pop_element = parent_card.find('div', class_='headlineNumber')
    rec.population_total = pop_element.text.strip().replace(',', '')
```

```python
# 1. Target the unique population density icon
density_img = soup.find('img', src=lambda x: x and 'icons/population_density.png' in x)

if density_img:
    # 2. Go to the card container and grab the correct headline number
    parent_card = density_img.find_parent('div', class_='card')
    density_element = parent_card.find('div', class_='headlineNumber')
    rec.population_density = density_element.text.strip().replace(',', '')
```
