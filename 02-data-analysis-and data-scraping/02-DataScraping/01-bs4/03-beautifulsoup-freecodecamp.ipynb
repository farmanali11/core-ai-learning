```python
from bs4 import BeautifulSoup
with open("home.html", "r") as html_file:
  content = html_file.read()
  soup = BeautifulSoup(content,'lxml')
print(soup.prettify())
```

```python
courses_html_tags = soup.find_all('h5')
for course in courses_html_tags:
  print(course.text)
```

```python
course_cards = soup.find_all('div',class_= 'card')

for course in course_cards:
  course_name = course.h5.text
  course_price = course.a.text.split()[-1]
  print(course_name)
  print(course_price)
```

```python
import requests
from bs4 import BeautifulSoup
url = 'https://timesjobs.com/job-search?keywords=%22python%22&location=&experience=&refreshed=true'
html_text = requests.get(url,verify=False).text

soup = BeautifulSoup(html_text,'lxml')
jobs = soup.find('a',class_ = 'absolute left-0 right-0 top-0 bottom-0 z-20')

print(jobs)
```
