## Find Missing Data 

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

## Removing Null Data

```python
df.dropna()
```

```python
df.dropna(thresh=3)
```

```python
df.fillna(0)
```

```python
values = {'0':0,'1':100,'2':200,'D':400}

df.fillna(value=values)
```

```python
df.fillna(df.mean())
```
