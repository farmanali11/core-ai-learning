# Bar Plot

```python
import matplotlib.pyplot as plt
x=['python','c','javascript','c++','java']
y =[89,78,67,78,80]
plt.xlabel("Languages")
plt.ylabel("No.")
plt.title("Popularity")
c=['r','b','y','g']
plt.bar(x,y,width=0.5,color=c,align='center',edgecolor='black',linewidth=6,alpha=0.5,label='Farman')
plt.legend()
plt.show()
```

# Double Bar Graph

```python
import matplotlib.pyplot as plt
import numpy as np

x = ["python", "c", "javascript", "c++", "java"]
y = [89, 78, 67, 78, 80]
z = [10, 50, 60, 30, 70]

width = 0.2
p = np.arange(len(x))

# Define two separate x-positions for the two charts
p1 = p
p2 = [j + width for j in p]

plt.xlabel("Languages")
plt.ylabel("No.")
plt.title("Popularity")

# Plot first bar chart at p1
plt.bar(p1, y, width=width, color="r", label="Farman")

# Plot second bar chart at p2
plt.bar(p2, z, width=width, color="y", label="Farman1")

# Adjust the x-ticks to be between the two bars
plt.xticks(p + width / 2, x)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = ["python", "c", "javascript", "c++", "java"]
y = [89, 78, 67, 78, 80]
z = [10, 50, 60, 30, 70]

width = 0.2
p = np.arange(len(x))
p1 = [j + width for j in p]
plt.xlabel("Languages")
plt.ylabel("No.")
plt.title("Popularity")

plt.bar(
    p1,
    y,
    color="r",
    label="Farman",
)
plt.barh(
    p1,
    z,
    color="y",
    label="Farman1",
)

plt.xticks(p + width / 2, x)
plt.legend()
plt.show()
```
