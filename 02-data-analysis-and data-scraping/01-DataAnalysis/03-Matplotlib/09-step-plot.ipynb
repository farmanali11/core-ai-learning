# **Step Plot** 

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]

area = [2,5,3,2,4]

plt.step(x,area,marker='o',color='r',ms=10,label="Python")
plt.grid()
plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel('y-axis')
plt.legend(loc=1)
plt.show()
```
