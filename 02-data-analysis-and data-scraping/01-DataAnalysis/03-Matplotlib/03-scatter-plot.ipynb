# Scatter Plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6,7]
y=[2,3,1,4,5,3,6]
colors=[10,20,30,40,50,60,70]
sizes = [100,200,400,500,200,800,700]
plt.scatter(x, y, c=colors, cmap='BrBG', s=sizes, marker="*", edgecolors='w')
plt.title("Scatter",fontsize=15)
plt.colorbar()
plt.xlabel("Day")
plt.ylabel("No")
plt.show()
```

# MultiScatter Graph

```python
x = [1, 2, 3, 4, 5, 6, 7]
y = [2, 3, 1, 4, 5, 3, 6]
y2=[2,4,6,2,4,2,4]
colors = [10, 20, 30, 40, 50, 60, 70]
sizes = [100, 200, 400, 500, 200, 800, 700]
plt.scatter(x, y, c=colors, cmap="BrBG", s=sizes, marker="*", edgecolors="w")
plt.scatter(x, y2, c=colors, cmap="BrBG", s=sizes, edgecolors="w")
plt.title("Scatter", fontsize=15)
plt.colorbar()
plt.xlabel("Day")
plt.ylabel("No")
plt.show()
```
