# **Area vs Stack**

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
l = ["area","area2","area3"]

area1=[1,3,2,5,4]

area2=[2,3,1,2,3]

area3=[4,2,4,4,1]


plt.stackplot(x,area1,area2,area3,labels=l)
plt.legend()
plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-label")
plt.show()
```
