```python
import numpy as np

import pandas as pd

import matplotlib.pyplot as plt
```

```python
x=[10,20,30,40]

y=["c","c++","Java","Python"]
ex=[0.2,0.0,0.0,0.0]


plt.pie(
    x,
    labels=y,
    explode=ex,
    autopct="%0.1f%%",
    shadow=True,
    radius=1.3,
    labeldistance=0.2,
    startangle=0,
    textprops={"fontsize":15},
    counterclock=False,
    wedgeprops={"edgecolor":"k","linewidth":1.5}
)

plt.show
```
